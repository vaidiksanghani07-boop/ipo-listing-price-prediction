```python
import pandas as pd
import numpy as np
import joblib

from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
```

```python
data = pd.read_csv("../data/data_preprossing.csv")

X = data.drop(["List Price","IPO_Name","Date","Listing Gain","CMP(BSE)","CMP(NSE)","Current Gains"], axis=1)

y = data["List Price"]

X_train, X_test, y_train, y_test = train_test_split(X,y,test_size=0.20,random_state=1)
```

```python
models = {
    "Linear Regression": joblib.load("../notebooks/linear_regression.pkl"),
    "Decision Tree": joblib.load("../notebooks/Decision_tree.pkl"),
    "Tuned Decision Tree": joblib.load("../notebooks/Decision_Tree_Hyperparameter_Tuning.pkl"),
    "Random Forest": joblib.load("../notebooks/random_forest.pkl"),
    "Tuned Random Forest": joblib.load("../notebooks/random_forest_Hyperparameter_Tunnig.pkl"),
    "XGBoost": joblib.load("../notebooks/Xgboost.pkl"),
    "Tuned XGBoost": joblib.load("../notebooks/Xgboost_Hyperparameter_Tuning.pkl")
}
```

```python
results = []

for name, model in models.items():

    y_pred = model.predict(X_test)

    mae = mean_absolute_error(y_test, y_pred)
    mse = mean_squared_error(y_test, y_pred)
    rmse = np.sqrt(mse)
    r2 = r2_score(y_test, y_pred)

    results.append({
        "Model": name,
        "MAE": mae,
        "MSE": mse,
        "RMSE": rmse,
        "R2": r2
    })

comparison = pd.DataFrame(results)

comparison
```
