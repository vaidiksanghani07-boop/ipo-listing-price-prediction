```python
import pandas as pd
import numpy as np

data1 = pd.read_excel("../data/Initial Public Offering.xlsx")
data1.shape
```

```python
data1.head()
```

```python
data2 = pd.read_excel("../data/Initial Public Offering (Updated).xlsx")
data2.shape
```

```python
data2.head()
```

```python
data = pd.concat([data1 ,data2] , ignore_index = True)
data.head(20)
```

```python
data.shape
```

```python
drop_data = ["Unnamed: 13", "Unnamed: 14", "Unnamed: 15","Unnamed: 16", "Unnamed: 17", "Unnamed: 18","Unnamed: 19", "Unnamed: 20"]

data.drop(drop_data, inplace=True, axis=1)
data.head(20)
```

```python
data.shape
```

```python
data.fillna(0, inplace=True)

print(data.isnull().sum())
```

```python
data[data.duplicated()]
```

```python
data["IPO_Name"].duplicated().sum()
```

```python
print(data.duplicated().sum())
data = data.drop_duplicates()
data = data.reset_index(drop=True)

print(data.shape)
```

```python
import matplotlib.pyplot as plt

top_20 = data.nlargest(20, "List Price")

plt.figure(figsize=(12, 6))

plt.bar(top_20["IPO_Name"], top_20["List Price"])

plt.xlabel("IPO Name")
plt.ylabel("List Price")
plt.title("Top 20 IPOs by Listing Price")

plt.xticks(rotation=90)
plt.tight_layout()

plt.show()
plt.show()
```

```python
data = data.to_csv("../data/data_preprossing.csv" , index = False)
```
